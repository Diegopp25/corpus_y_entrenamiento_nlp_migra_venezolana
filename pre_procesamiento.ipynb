```python
import pandas as pd
```

```python

# Leer archivos
df_espectador = pd.read_csv(
    "el_espectador_migracion_venezolana.csv",
    encoding="utf-8-sig"
)

df_heraldo = pd.read_csv(
    "elheraldo_migracion_venezolana.csv",
    encoding="utf-8-sig"
)

# Renombrar columnas de El Heraldo para que coincidan
df_heraldo = df_heraldo.rename(
    columns={
        "titulo": "Titulo",
        "url": "URL",
        "texto": "Cuerpo",
        "fecha": "Fecha",
        "autor": "Autor",
        "medio": "Medio",
        "pais": "Pais"
    }
)

# Mantener exactamente el mismo orden de columnas
columnas = [
    "Titulo",
    "URL",
    "Cuerpo",
    "Fecha",
    "Autor",
    "Medio",
    "Pais"
]

df_heraldo = df_heraldo[columnas]
df_espectador = df_espectador[columnas]

# Unir
df_final = pd.concat(
    [df_espectador, df_heraldo],
    ignore_index=True
)

# Opcional: eliminar duplicados por URL
df_final = df_final.drop_duplicates(
    subset=["URL"]
)

# Guardar
df_final.to_csv(
    "migracion_venezolana_noticias.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Noticias totales:", len(df_final))
print("Archivo guardado: migracion_venezolana_noticias.csv")
print(df_final.head())
```
