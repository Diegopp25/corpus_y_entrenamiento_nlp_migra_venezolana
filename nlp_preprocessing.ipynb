# Preprocesamiento 

#### Cargamos el modelo en español de spacy para preprocesamiento de texto

```python
!python -m spacy download es_core_news_sm
```

```python
# Cargar modelo en español
nlp = spacy.load('es_core_news_sm')
```

```python
# Probando spacy jijijiji

texto = "Hace mucho tiempo, frente el pelotón de fusilamiento, el coronel Aureliano Buendía habría de recordar aquella remota tarde en la que su padre lo llevo a conocer el hielo"
doc = nlp(texto)

for ent in doc.ents:
    print(ent.text, ent.label_)
```
